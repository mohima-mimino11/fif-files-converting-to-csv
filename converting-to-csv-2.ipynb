# **RAW-CSV-128-CHANNEL**

```python
import mne
import numpy as np
import pandas as pd
import os
import warnings

# Directory
input_dir = '/kaggle/input/preproccesed-dataset/128-channel-ERP/kaggle/working/preprocessed_raw_output_directory'
output_dir = '/kaggle/working/csv_files_from_raw'
os.makedirs(output_dir, exist_ok=True) 

# Function to convert .fif files to CSV
def convert_fif_to_csv(file_path, output_dir):
    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        try:
            epochs = mne.read_epochs(file_path, verbose=False)
            if len(epochs) == 0:
                print(f"No data in file {file_path}, skipping.")
                return
            df_epochs = epochs.to_data_frame()
            csv_file_path = os.path.join(output_dir, os.path.basename(file_path).replace('-epo.fif', '.csv'))
            df_epochs.to_csv(csv_file_path, index=False)
            print(f"Converted {file_path} to {csv_file_path}")
        except Exception as e:
            print(f"Error processing {file_path}: {e}")

# Iterate through the input directory
for file in os.listdir(input_dir):
    file_path = os.path.join(input_dir, file)
    if file.endswith('-epo.fif'):
        convert_fif_to_csv(file_path, output_dir)
```
